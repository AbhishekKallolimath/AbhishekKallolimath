```python
# Install required packages (run this cell first)
!pip install nltk spacy
!python -m spacy download en_core_web_sm

import nltk
import spacy
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk import pos_tag
import os

nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')

nlp = spacy.load("en_core_web_sm")

def count_pos_nltk(tokens_tags):
    nouns = verbs = adj = adv = 0
        for word, tag in tokens_tags:
                if tag.startswith('NN'):   nouns += 1
                        elif tag.startswith('VB'): verbs += 1
                                elif tag.startswith('JJ'): adj += 1
                                        elif tag.startswith('RB'): adv += 1
                                            return nouns, verbs, adj, adv

                                            def count_pos_spacy(doc):
                                                nouns = verbs = adj = adv = 0
                                                    for token in doc:
                                                            if token.pos_ in ('NOUN', 'PROPN'): nouns += 1
                                                                    elif token.pos_ == 'VERB': verbs += 1
                                                                            elif token.pos_ == 'ADJ': adj += 1
                                                                                    elif token.pos_ == 'ADV': adv += 1
                                                                                        return nouns, verbs, adj, adv

                                                                                        def process_file(text):
                                                                                            sentences = sent_tokenize(text)
                                                                                                words = word_tokenize(text)
                                                                                                    tokens_tags = pos_tag(words)
                                                                                                        n_nltk, v_nltk, adj_nltk, adv_nltk = count_pos_nltk(tokens_tags)
                                                                                                            doc = nlp(text)
                                                                                                                n_spacy, v_spacy, adj_spacy, adv_spacy = count_pos_spacy(doc)
                                                                                                                    return {
                                                                                                                            'sentences': len(sentences),
                                                                                                                                    'words': len(words),
                                                                                                                                            'nltk_nouns': n_nltk,
                                                                                                                                                    'nltk_verbs': v_nltk,
                                                                                                                                                            'nltk_adj': adj_nltk,
                                                                                                                                                                    'nltk_adv': adv_nltk,
                                                                                                                                                                            'spacy_nouns': n_spacy,
                                                                                                                                                                                    'spacy_verbs': v_spacy,
                                                                                                                                                                                            'spacy_adj': adj_spacy,
                                                                                                                                                                                                    'spacy_adv': adv_spacy
                                                                                                                                                                                                        }

                                                                                                                                                                                                        # Replace with your actual roll number
                                                                                                                                                                                                        roll_number = "20241COM1661"

                                                                                                                                                                                                        # List of file names as they appear in your zip (adjust if different)
                                                                                                                                                                                                        files = ["File1.txt", "File2.txt", "File3.txt", "File4.txt"]

                                                                                                                                                                                                        output_lines = [roll_number, ""]

                                                                                                                                                                                                        for fname in files:
                                                                                                                                                                                                            try:
                                                                                                                                                                                                                    with open(fname, 'r', encoding='utf-8') as f:
                                                                                                                                                                                                                                text = f.read()
                                                                                                                                                                                                                                        stats = process_file(text)
                                                                                                                                                                                                                                                output_lines.append(f"{fname}:")
                                                                                                                                                                                                                                                        output_lines.append(f"Sentences = {stats['sentences']}")
                                                                                                                                                                                                                                                                output_lines.append(f"Words = {stats['words']}")
                                                                                                                                                                                                                                                                        output_lines.append(f"NLTK Nouns = {stats['nltk_nouns']}")
                                                                                                                                                                                                                                                                                output_lines.append(f"NLTK Verbs = {stats['nltk_verbs']}")
                                                                                                                                                                                                                                                                                        output_lines.append(f"NLTK Adjectives = {stats['nltk_adj']}")
                                                                                                                                                                                                                                                                                                output_lines.append(f"NLTK Adverbs = {stats['nltk_adv']}")
                                                                                                                                                                                                                                                                                                        output_lines.append(f"Spacy Nouns = {stats['spacy_nouns']}")
                                                                                                                                                                                                                                                                                                                output_lines.append(f"Spacy Verbs = {stats['spacy_verbs']}")
                                                                                                                                                                                                                                                                                                                        output_lines.append(f"Spacy Adjectives = {stats['spacy_adj']}")
                                                                                                                                                                                                                                                                                                                                output_lines.append(f"Spacy Adverbs = {stats['spacy_adv']}")
                                                                                                                                                                                                                                                                                                                                        output_lines.append("")
                                                                                                                                                                                                                                                                                                                                            except FileNotFoundError:
                                                                                                                                                                                                                                                                                                                                                    output_lines.append(f"{fname}: File not found. Please upload the file to Colab.")
                                                                                                                                                                                                                                                                                                                                                            output_lines.append("")

                                                                                                                                                                                                                                                                                                                                                            # Print the result
                                                                                                                                                                                                                                                                                                                                                            print("\n".join(output_lines))

                                                                                                                                                                                                                                                                                                                                                            # Optionally, save to a file
                                                                                                                                                                                                                                                                                                                                                            with open("output.txt", "w", encoding='utf-8') as out:
                                                                                                                                                                                                                                                                                                                                                                out.write("\n".join(output_lines))
```
